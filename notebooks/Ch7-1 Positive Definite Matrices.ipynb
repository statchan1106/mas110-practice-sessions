## Notation

- $x^\top Ax$ · The lecture writes $x$ as a column vector. The code stores it as a one-dimensional array, so `x @ Ax` takes the dot product that represents multiplying by $x^\top$.

- $A=R^\top R$ · The lecture uses upper triangular $R$ with positive diagonal entries. NumPy defaults to the lower triangular factor $L$, so the code sets $R=L^\top$.

## 1. Multiply first, then take a dot product

The parentheses in $x^\top(Ax)$ give the order of the two calculations. Each row of $A$ supplies one coordinate of $Ax$.

**Predict first.** What are the two entries of $Ax$? What number do you get after the dot product?

```python
import numpy as np
A = np.array([[5., 4.], [4., 5.]])
x = np.array([1., -1.])
Ax = A @ x
quadratic = x @ Ax
print("Ax =", Ax, "\nQuadratic form =", quadratic)
```

The intermediate vector has two entries. The dot product adds two coordinate products.

$$x^\top(Ax)=1\cdot1+(-1)(-1)=2$$

## 2. Match NumPy’s factor to the lecture

The default Cholesky call gives lower triangular $L$ with $A=LL^\top$. Set $R=L^\top$ to read the product in lecture notation.

**Predict first.** After transposing $L$, which off-diagonal corner of $R$ contains the zero?

```python
L = np.linalg.cholesky(A)
R = L.T
rebuilt = R.T @ R
rebuild_ok = np.allclose(rebuilt, A)
print("L =\n", L, "\nR =\n", R, "\nRebuilt A =\n", rebuilt, "\nMatches A:", rebuild_ok)
```

The factor entries are rounded here. The code uses full precision when multiplying them.

$$A=R^\top R$$

## 3. Compute the squared length

Using $A=R^\top R$ gives $x^\top Ax=(Rx)^\top(Rx)$. We already know the value on the left, so now compute the one on the right.

**Predict first.** What value should the two squares add to, even though one entry of $Rx$ is negative?

```python
Rx = R @ x
squared_length = Rx @ Rx
length_ok = np.isclose(squared_length, quadratic)
print("Rx =", Rx, "\nSquared length =", squared_length, "\nMatches the quadratic form:", length_ok)
```

The two entries of $Rx$ have opposite signs. Squaring them produces two positive contributions.

$$|Rx|^2=\tfrac{1}{5}+\tfrac{9}{5}=2$$

## Verify the original inputs

Run these checks before changing anything.

```python
assert np.allclose(Ax, [1, -1])
assert np.isclose(quadratic, 2)
assert np.allclose(R, np.array([[5, 4], [0, 3]]) / np.sqrt(5))
assert rebuild_ok and length_ok
```
