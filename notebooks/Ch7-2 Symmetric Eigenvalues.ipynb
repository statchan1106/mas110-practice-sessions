## Notation

- $Av_i=\lambda_i(A)v_i$ · The lecture uses descending order. After reversing the NumPy output, column `V[:, 0]` is $v_1$ and `lam[0]` is $\lambda_1(A)$.

- $R_A(x)$ · For $x\ne0$, divide $x^\top Ax$ by $x^\top x$. In the lecture proof, $x_i=v_i^\top x$ is a coordinate in the eigenbasis. The code stores those coordinates in `coords`.

## 1. Order the eigenpairs as in the lecture

NumPy returns the eigenvalues from smallest to largest. Each eigenvector is a column paired with the eigenvalue at the same index.

**Predict first.** To move the larger eigenvalue to index 0, which axis of the eigenvector array must you reverse?

```python
import numpy as np
A = np.array([[5., 4.], [4., 5.]])
lam_asc, V_asc = np.linalg.eigh(A)
lam = lam_asc[::-1]
V = V_asc[:, ::-1]
print("Ascending =", lam_asc, "\nDescending =", lam, "\nV =\n", V)
```

The first column points along $(1,1)^\top$ and the second along $(1,-1)^\top$, up to signs.

$$\lambda_1(A)=9,\qquad \lambda_2(A)=1$$

## 2. Read one Rayleigh quotient as an average

The dot products with the columns of $V$ give the eigenbasis coordinates. Square and normalize those coordinates to get a weight for each eigenvalue.

**Predict first.** The vector $(1,0)^\top$ has equal absolute coordinates along the two eigendirections. What average do you expect?

```python
x = np.array([1., 0.])
coords = V.T @ x
weights = coords**2 / (coords @ coords)
rayleigh = weights @ lam
direct = (x @ A @ x) / (x @ x)
print("Eigenbasis coordinates =", coords, "\nWeights =", weights, "\nRayleigh quotient =", rayleigh, "\nDirect =", direct)
```

Both eigendirections carry half the squared length of $x$. Their two contributions give $R_A(x)=5$.

$$R_A(e_1)=\tfrac12\cdot9+\tfrac12\cdot1=5$$

## 3. Add a rank-one matrix and compare the spectra

The outer product of $q=e_1$ with itself adds one to the top-left entry. We only need the new eigenvalues, so the code uses `eigvalsh`.

**Predict first.** Will the smaller eigenvalue of $A+qq^\top$ stay between the two old eigenvalues?

```python
q = np.array([1., 0.])
B = np.outer(q, q)
lam_updated = np.linalg.eigvalsh(A + B)[::-1]
interlaces = lam_updated[0] >= lam[0] >= lam_updated[1] >= lam[1]
print("B =\n", B, "\nUpdated eigenvalues =", lam_updated, "\nInterlaces:", interlaces)
```

The two eigenvalues both increase. The new smaller eigenvalue is still below the old larger one, so `interlaces` is `True`.

$$9.531>9>1.469>1$$

## Verify the original inputs

Run these checks before changing anything.

```python
assert np.allclose(lam_asc, [1, 9]) and np.allclose(lam, [9, 1])
assert np.allclose(A @ V, V * lam)
assert np.allclose(weights, [0.5, 0.5]) and np.allclose([rayleigh, direct], [5, 5])
assert np.allclose(lam_updated, [(11 + np.sqrt(65)) / 2, (11 - np.sqrt(65)) / 2])
assert interlaces
```
